# The Pillow library contains all the basic image processing functionality. You can do image resizing, rotation and transformation.
## Pillow module allows you to pull some statistics data out of image using histogram method, which later can be used for statistical analysis and automatic contrast enhancement.

```python
from PIL import Image
import PIL
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
img1 = Image.open("Cars.jpg")
img2 = Image.open("images.jpg")
```

```python
plt.imshow(img1)
plt.show()
```

# Resize image

```python
img_re = img2.resize((1200,1000))
plt.imshow(img_re)
plt.show()
```

```python
img2_resized = img2.resize((1200, 800))
print("Original image size: {}".format(img2.size))
print("Resized image size: {}".format(img2_resized.size))
plt.imshow(img2_resized)
plt.show()
```

# Horizontal concatenation

```python
result_img = Image.new('RGB', (img1.width + img2.width, min(img1.height, img2.height)))
result_img.paste(img1, (0, 0))
result_img.paste(img2, (img1.width, 0))
plt.imshow(result_img)
plt.show()
```

# Vertical concatenation

```python
result_img = PIL.Image.new('RGB',(min(img1.width, img2.width), img1.height + img2.height))
result_img.paste(img1, (0, 0))
result_img.paste(img2, (0, img1.height))
plt.imshow(result_img)
plt.show()
```

```python
result_img.save("new_image.jpg")
```
